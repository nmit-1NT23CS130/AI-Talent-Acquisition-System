```python
from datasets import load_dataset
import pandas as pd

print(" Libraries imported successfully!")
```

```python
ds = load_dataset("cnamuangtoun/resume-job-description-fit")

print(" Dataset loaded!")
print(ds)
```

```python
df_train = pd.DataFrame(ds['train'])
df_test = pd.DataFrame(ds['test'])

print("Train shape:", df_train.shape)
print("Test shape:", df_test.shape)
print("\nColumns:", df_train.columns.tolist())
```

```python
# See label distribution
print("=== LABEL COUNTS ===")
print(df_train['label'].value_counts())

# See first row clearly
print("\n=== SAMPLE RESUME (first 300 chars) ===")
print(df_train['resume_text'][0][:300])

print("\n=== SAMPLE JD (first 300 chars) ===")
print(df_train['job_description_text'][0][:300])

print("\n=== LABEL ===")
print(df_train['label'][0])
```

```python
# Total rows
print("=== DATASET SIZE ===")
print(f"Total pairs:     {len(df_train) + len(df_test)}")
print(f"Train pairs:     {len(df_train)}")
print(f"Test pairs:      {len(df_test)}")

# Unique resumes and JDs
print("\n=== UNIQUE COUNTS ===")
print(f"Unique Resumes:  {df_train['resume_text'].nunique()}")
print(f"Unique JDs:      {df_train['job_description_text'].nunique()}")

# Label distribution
print("\n=== LABEL COUNTS ===")
print(df_train['label'].value_counts())
```

```python
df_train.to_csv("../data/resume_jd_train.csv", index=False)
df_test.to_csv("../data/resume_jd_test.csv", index=False)

print("Saved to data folder!")
print("Train:", df_train.shape)
print("Test:", df_test.shape)
```
